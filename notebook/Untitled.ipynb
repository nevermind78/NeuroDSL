```python
def permutations(elements):
    if len(elements) <= 1:
        return [elements]
    
    all_perms = []
    for i in range(len(elements)):
        first = elements[i]
        remaining_elements = elements[:i] + elements[i+1:]
        for p in permutations(remaining_elements):
            all_perms.append([first] + p)
    
    return all_perms

# Example usage:
elements = [1, 2, 3,4]
perms = permutations(elements)
for perm in perms:
    print(perm)
```
